# Prolog-RAG: Basic Demo
This section demonstrates the core capabilities of the Prolog-RAG system in isolation.

```python
from prolog_rag_project.core.prolog_rag import PrologRAG
import logging

# Set up logging
logging.basicConfig(level=logging.INFO)

print("Initializing system pipeline...")
rag = PrologRAG()
print("Prolog-RAG is ready!")
```

```python
question = "What is the profit margin for 2023?"
result = rag.query(question)

print(f"Question: {result['question']}")
print(f"Route Type: {result['route']}")
print(f"Answer: {result['answer']}")
```

```python
print("--- Proof Trace ---")
if result['proof_trace']:
    for line in result['proof_trace']:
        print(f"  -> {line}")
else:
    print("  (No formal proof trace available in current system release)")
```

```python
import matplotlib.pyplot as plt

methods = ['Prolog-RAG', 'Traditional RAG']
values = [1, 0]

plt.figure(figsize=(8, 5))
bars = plt.bar(methods, values, color=['#4285F4', '#EA4335'])
plt.title('Explainability Comparison (Formal Logic Proof Trace)')
plt.ylabel('Has Proof (Boolean)')
plt.ylim(0, 1.2)

for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval + 0.05, 'Yes' if yval == 1 else 'No', ha='center', va='bottom', fontsize=12, fontweight='bold')

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

# Prolog-RAG: Industry Standard Comparison Demo
This section demonstrates the real-time performance difference between the **Prolog-RAG** symbolic approach and multiple SOTA baselines.

```python
import os
import sys
import pandas as pd
from IPython.display import display, Markdown, HTML

# Import all systems
from prolog_rag_project.core.prolog_rag import PrologRAG
from prolog_rag_project.baselines.naive_rag import NaiveRAG
from prolog_rag_project.baselines.graph_rag import SOTAGraphRAG
from prolog_rag_project.baselines.corrective_rag import CorrectiveRAG
from prolog_rag_project.baselines.contextual_rag import ContextualRAG

print("✅ All RAG systems loaded successfully!")
```

```python
question_arena = "What was the total allocated cost for management support, shared services, and benefits for the year ended December 31, 2011?"

# Initialize systems
print("Initializing all systems (this may take a moment)...")
prolog_arena = PrologRAG()
naive_arena = NaiveRAG()
graph_arena = SOTAGraphRAG()
crag_arena = CorrectiveRAG()
contextual_arena = ContextualRAG()

print("Running queries across the arena...")
results_arena = {
    "Prolog-RAG": prolog_arena.query(question_arena),
    "Naive RAG": naive_arena.query(question_arena),
    "Graph RAG": graph_arena.query(question_arena),
    "CRAG": crag_arena.query(question_arena),
    "Contextual RAG": contextual_arena.query(question_arena)
}
```

```python
html_str = """
<style>
    .comparison-table { width: 100%; border-collapse: collapse; font-family: sans-serif; }
    .comparison-table th { background-color: #f2f2f2; padding: 12px; text-align: left; border: 1px solid #ddd; }
    .comparison-table td { padding: 12px; border: 1px solid #ddd; vertical-align: top; }
    .prolog-row { background-color: #e8f5e9; font-weight: bold; }
    .badge { padding: 4px 8px; border-radius: 4px; color: white;}
    .badge-prolog { background-color: #2e7d32; }
    .badge-vector { background-color: #1565c0; }
</style>
<table class='comparison-table'>
    <tr>
        <th>System</th>
        <th>Type</th>
        <th>Answer Summary</th>
        <th>Proof Trace (Explainability)</th>
    </tr>
"""

for name, res in results_arena.items():
    row_class = "prolog-row" if name == "Prolog-RAG" else ""
    badge_class = "badge-prolog" if name == "Prolog-RAG" else "badge-vector"
    route = res.get('route', 'VECTOR')
    proof = "<br>".join(res.get('proof_trace', [])) if res.get('proof_trace') else "<i>(No formal proof trace available)</i>"
    
    html_str += f"""
    <tr class='{row_class}'>
        <td>{name}</td>
        <td><span class='badge {badge_class}'>{route}</span></td>
        <td>{res.get('answer', '')[:250]}...</td>
        <td>{proof}</td>
    </tr>
    """

html_str += "</table>"
display(HTML(html_str))
```
